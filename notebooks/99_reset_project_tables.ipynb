# 99 — Reset Project Tables

Use only to replay the demonstration from scratch. This drops the project's Bronze, Silver, and Gold tables but does not delete the uploaded batch files in the volume.

```python
dbutils.widgets.text("catalog", "workspace")
dbutils.widgets.text("schema", "default")
CATALOG = dbutils.widgets.get("catalog")
SCHEMA = dbutils.widgets.get("schema")

tables = [
    "bronze_taxi_raw", "silver_taxi_trips", "gold_hourly_demand_by_cell",
    "gold_weekly_fleet_kpis", "gold_fare_anomaly_review", "gold_genie_taxi_summary"
]
for t in tables:
    spark.sql(f"DROP TABLE IF EXISTS `{CATALOG}`.`{SCHEMA}`.`{t}`")
    print("Dropped", f"{CATALOG}.{SCHEMA}.{t}")
```
