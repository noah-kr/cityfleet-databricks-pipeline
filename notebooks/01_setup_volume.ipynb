# 01 — CityFleet Taxi Project Setup

Creates the Unity Catalog volume folders used by the prototype. Run this once before uploading the taxi batch files.

Default target used throughout the project:

- Catalog: `workspace`
- Schema: `default`
- Volume: `cityfleet_taxi_project`
- Batch landing folder: `/Volumes/workspace/default/cityfleet_taxi_project/landing/`

After this notebook finishes, upload `sample_data/taxi_batches/batch_001.csv.gz`, `batch_002.csv.gz`, and `batch_003.csv.gz` into the landing folder using the Databricks UI.

```python
# Parameters: change only if your workspace uses different names.
dbutils.widgets.text("catalog", "workspace")
dbutils.widgets.text("schema", "default")
dbutils.widgets.text("volume", "cityfleet_taxi_project")

CATALOG = dbutils.widgets.get("catalog")
SCHEMA = dbutils.widgets.get("schema")
VOLUME = dbutils.widgets.get("volume")

volume_path = f"/Volumes/{CATALOG}/{SCHEMA}/{VOLUME}"
landing_dir = f"{volume_path}/landing"
checkpoint_dir = f"{volume_path}/checkpoints"
print("Volume path:", volume_path)
print("Landing folder:", landing_dir)
```

```python
# Create schema and managed volume
spark.sql(f"CREATE SCHEMA IF NOT EXISTS `{CATALOG}`.`{SCHEMA}`")
spark.sql(f"CREATE VOLUME IF NOT EXISTS `{CATALOG}`.`{SCHEMA}`.`{VOLUME}`")

dbutils.fs.mkdirs(landing_dir)
dbutils.fs.mkdirs(checkpoint_dir)
dbutils.fs.mkdirs(f"{volume_path}/reports")

print("Setup complete. Upload the three batch_*.csv.gz files to:")
print(landing_dir)
```
